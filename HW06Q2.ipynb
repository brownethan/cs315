```python
import random
import time

def quickselect(A, i):

    if len(A) == 1:
        return A[0]

    pivot = random.choice(A)
    lows = [x for x in A if x < pivot]
    highs = [x for x in A if x > pivot]
    pivots = [x for x in A if x == pivot]

    if i <= len(lows):
        return quickselect(lows, i)

    elif i <= len(lows) + len(pivots):
        return pivot

    else:
        new_i = i - len(lows) - len(pivots)
        return quickselect(highs, new_i)


# --- Testing Functions and Asymptotic Running Time Confirmation ---

def run_test(name, data, k):
    """
    Runs quickselect on the data, measures the time, and checks for correctness.
    """
    start_time = time.time()

    # Use the quickselect function to find the k-th smallest element
    result = quickselect(data, k)

    end_time = time.time()

    # Validation (using standard sort for expected result)
    expected = sorted(data)[k - 1]

    print(f"\n{name}")
    print(f"List Size (N): {len(data):,}")
    print(f"Target (i): {k}-th smallest element")
    print(f"Time taken: {end_time - start_time:.6f} seconds")
    print(f"Result Correct: {result == expected}")


# --- Test Cases for Variety of Data Orderings ---

# 1. Randomly Ordered Data (Average Case)
N_RANDOM = 100_000
A_random = [random.randint(1, N_RANDOM) for _ in range(N_RANDOM)]
run_test("Test 1: Random Data (Median)", A_random, N_RANDOM // 2)

# 2. Fully Sorted Data (A common test case for selection algorithms)
N_SORTED = 100_000
A_sorted = list(range(1, N_SORTED + 1))
run_test("Test 2: Fully Sorted Data (75th percentile)", A_sorted, N_SORTED * 3 // 4)

# 3. Data with many Duplicates
N_DUPLICATES = 100_000
A_duplicates = [random.choice([1, 5, 10, 50, 100]) for _ in range(N_DUPLICATES)]
run_test("Test 3: Duplicate Data (Smallest element)", A_duplicates, 1)


# --- Asymptotic Running Time (O(N) Confirmation) ---

# Compare time for N and 2N to confirm linear time complexity (O(N)).
N_LINEAR_SMALL = 250_000
N_LINEAR_LARGE = 500_000

# Generating large random lists for O(N) testing
A_linear_small = [random.randint(1, N_LINEAR_SMALL) for _ in range(N_LINEAR_SMALL)]
A_linear_large = [random.randint(1, N_LINEAR_LARGE) for _ in range(N_LINEAR_LARGE)]

print("\n--- Asymptotic Running Time Confirmation (O(N)) ---")
run_test(f"O(N) Check: N={N_LINEAR_SMALL:,}", A_linear_small, N_LINEAR_SMALL // 2)
run_test(f"O(N) Check: N={N_LINEAR_LARGE:,}", A_linear_large, N_LINEAR_LARGE // 2)

print("\n \n Worst case: O(N^2)")
```
