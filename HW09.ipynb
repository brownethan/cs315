```python
import numpy as np
```

```python
input_string = "NARRFSECCNTCXAIOUITEPSSIISHHEOSAEIAK"

matrix = np.array([])

for char in input_string:
    matrix = np.append(matrix, char)

rows_cols = int(len(input_string)**0.5) # Calculate rows_cols based on string length
matrix = matrix.reshape(rows_cols, rows_cols)

print("1. Caesar Square Decryption Setup (Matrix):")
print(matrix)
```
