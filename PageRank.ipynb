Google's Simple PageRank

```python
import numpy as np
```

```python
A = np.array( ([0,0,1],
               [0.5,0,0],
               [0.5,1,0]) )
print(A)
```

```python
eigenvalues, eigenvectors = np.linalg.eig(A)
print("Eigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```

```python
eigenvector_1 = eigenvectors[:, 0]
print(eigenvector_1)
```

```python
l1_norm = np.linalg.norm(eigenvector_1, 1)
print(l1_norm)
scaled_ev = eigenvector_1 / l1_norm
print(scaled_ev)
```

```python
print(A @ scaled_ev)
```

```python
iter_vec=np.array([1,1,1])
print(iter_vec)
```

```python
iter_vec = A @ iter_vec
l1_norm = np.linalg.norm(iter_vec, 1)
iter_vec = iter_vec / l1_norm
print(iter_vec)
```
